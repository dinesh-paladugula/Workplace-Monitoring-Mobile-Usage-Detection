```python
from ultralytics import YOLO

# Load pre-trained YOLOv8 model
model = YOLO("yolov8n.pt")  # 'n' = nano, for speed. Use 's' for better accuracy.

# Train the model on your dataset
model.train(data=r"/content/drive/MyDrive/Workplace Monitoring System/Dataset/data.yaml", epochs=50, imgsz=640, batch=16, device="cuda")  # Use 'cuda' for GPU
```

```python
import cv2
from ultralytics import YOLO

# Load trained YOLOv8 model
model = YOLO("/content/drive/MyDrive/Workplace Monitoring System/best.pt")

video_path = "/content/drive/MyDrive/Workplace Monitoring System/Test Data /test2.mp4"
cap = cv2.VideoCapture(video_path)

# Get video properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Define output path and video writer
output_path = "/content/drive/MyDrive/Workplace Monitoring System/Output Data/output_video1.mp4"
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

# Process the video
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLOv8 inference on the frame
    results = model(frame)

    # Count "mobile usage detected"
    mobile_count = 0
    for result in results:
        for box in result.boxes:
            class_id = int(box.cls[0])  # Get class ID
            confidence = box.conf[0]  # Confidence score
            if class_id == 0 and confidence > 0.5:  # Adjust confidence threshold
                mobile_count += 1
                x1, y1, x2, y2 = map(int, box.xyxy[0])  # Bounding box
                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)  # Draw box
                cv2.putText(frame, "Mobile Usage", (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    # Display alert if mobile usage is detected
    if mobile_count > 0:
        alert_text = f"{mobile_count} people were using mobile - ALERT"
        cv2.putText(frame, alert_text, (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)

    # Write frame to output video
    out.write(frame)

# Release resources
cap.release()
out.release()
cv2.destroyAllWindows()

print(f"✅ Processed video saved at: {output_path}")
```
